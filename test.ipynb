```python
import numpy as np
import matplotlib.pyplot as plt

# Some task specific data
lvl = 83 # Agility lvl

# Probability of success per trial
denominator = 9779 - 25 * lvl  # Penguin
p = 1 / denominator

# Number of trials range (enough to show the distribution clearly)
trials = np.arange(0, 30_000)
total_trials = 500 # Laps completed

# CDF of geometric distribution
cdf = 1 - (1 - p) ** trials

# Find the trial number where cumulative probability reaches ~50%
k_50 = np.argmin(np.abs(cdf - 0.5))
cdf_50 = cdf[k_50]

# Plot
plt.figure(figsize=(8,5))
plt.plot(trials, cdf, label="CDF")
plt.axvline(k_50, color="green", linestyle="--", label=f"50% point ≈ {k_50}")
plt.axvline(total_trials, color="red", linestyle="--", label=f"Current point ({total_trials}) ≈ {cdf[total_trials] * 100:.2f}%")
plt.scatter([k_50], [cdf_50], color="black", zorder=5)
plt.scatter([total_trials], [cdf[total_trials]], color="black", zorder=5)

plt.title(f"CDF of Geometric Distribution (p=1/{denominator})")
plt.xlabel("Number of Trials")
plt.ylabel("Cumulative Probability")
plt.legend()
plt.grid(True)
plt.show()

```
